# Exemplo 3: Classificação - Decision Tree (breast Cancer)


# Dados

```python
from sklearn.datasets import load_breast_cancer

dados = load_breast_cancer()
```

```python
# Descrição da base de dados
print(dados.DESCR)
```

```python
import pandas as pd

# Atributos / Características (Resultados dos exames) 
X = pd.DataFrame(dados.data, columns=dados.feature_names)
X.info()
```

```python
# Variável alvo(Cancer/Não-cancer)
y = pd.Series(dados.target, name='cancer')
y.value_counts(normalize=True)
```

# Divisão treinamento/teste

```python
from sklearn.model_selection import train_test_split

X_treino, X_teste, y_treino, y_teste = train_test_split(
    X,
    y,
    test_size=0.007,
    random_state=40,
    stratify=y
)
```

```python
X_treino.shape, X_teste.shape, y_treino.shape, y_teste.shape
```

# Modelo

```python
from sklearn.tree import DecisionTreeClassifier, plot_tree

# Modelo                                    "random_state" é um cauculo de aleatoriedade
modelo = DecisionTreeClassifier(max_depth=3, random_state=42)
```

```python
# Treinamento        (Se fosse um algoritimo não supervisionado nós so colocariamos o X_treino)
modelo.fit(X_treino, y_treino)
```

# Previsão

```python
previsoes = modelo.predict(X_teste)
```

```python
previsoes
```

# Avaliação 

```python
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
acc = accuracy_score(y_teste, previsoes) * 100
print(f'O modelo acertou {acc:.2f}% das previsões.')
```

```python
print(confusion_matrix(y_teste, previsoes))
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(18,10))
plot_tree(modelo,
          feature_names=dados.feature_names,
          class_names=dados.target_names,
          filled=True,
          rounded=True,
          fontsize=8)

plt.show()
```
